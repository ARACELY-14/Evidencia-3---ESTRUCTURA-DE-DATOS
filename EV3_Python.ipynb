```python
import csv
```

```python
ventas = [
    ['Producto', 'Cantidad', 'Precio Unitario'],
    ['laptop', 5,150000],
    ['mouse',10,5000],
    ['Teclado',7,10000],
    ['Monitor',3,3000],
    ['Impresora',2,25000]
]
```

```python
with open('ventas.csv', 'w', newline='', encoding = 'utf-8') as archivo:
    escritor = csv.writer(archivo)
    escritor.writerows(ventas)
print("Archivo CSV creado exitosamente.")
```

```python
ventas_por_producto = {}
```

```python
with open('ventas.csv', 'w', newline='') as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(['Producto', 'Cantidad', 'Precio Unitario'])
    for producto, cantidad, precio in ventas[1:]:
        if producto in ventas_por_producto:
            ventas_por_producto[producto] ['Cantidad'] += cantidad
            ventas_por_producto[producto] ['Precio Unitario'] = precio
        else:
            ventas_por_producto[producto] = {'Cantidad': cantidad, 'Precio Unitario': precio}
    for producto, datos in ventas_por_producto.items():
        escritor.writerow([producto, datos['Cantidad'], datos['Precio Unitario']])
```

```python
with open('ventas.csv', 'r', newline='', encoding = 'utf-8') as archivo:
    lector = csv.reader(archivo)
    for fila in lector:
        print(fila)
```
